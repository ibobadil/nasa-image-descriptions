```python
csv_data = 'nasa_data_keywords.csv'
column_data = 'keywords'
df = pd.read_csv(csv_data)

keywords_text = ','.join(df[column_data].astype(str).tolist())

print(keywords_text)
```

```python
keyword_text_plot = keywords_text.replace(".","")
```

```python
keyword_text_plot
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

text = keyword_text_plot

wordcloud = WordCloud(width=2500, height=2000, background_color='white', min_font_size=15).generate(text)


plt.figure(figsize=(15,10))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.show()

wordcloud.to_file('nasa_wordcloud_square3.png')
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
```

```python
G = nx.from_pandas_edgelist(data_text, source='nasa_id', target='keywords')

options = {
    "font_size": 5,
    "node_color": "skyblue",
    "edge_color": "grey",
    "linewidths": 2,
    "width": 0.2,
}

pos_spaced = nx.spring_layout(G, scale=30, k=5)
plt.figure(figsize=(12,12))
plt.axis('off')
nx.draw(G, node_size=15, with_labels = True, **options)
plt.title('NASA Media and Keyword Connections')

plt.savefig('initial_attempt.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(25, 25))

g = nx.from_pandas_edgelist(data_text, source='nasa_id', target='keywords')
layout = nx.spring_layout(g, k=0.7, iterations=80)

item_size = [g.degree(keyword) * 30 for keyword in keywords]
nx.draw_networkx_nodes(g, layout, nodelist=keywords, node_size=item_size, node_color='powderblue')

nx.draw_networkx_nodes(g, layout, nodelist=ids, node_color='mediumaquamarine', node_size=100)

nx.draw_networkx_edges(g, layout, width=1, edge_color="black", alpha=0.2)

node_labels = dict(zip(keywords, keywords))
nx.draw_networkx_labels(g, layout, labels=node_labels, font_size=10)

for x in range(50):
    for nodex in g:
      for nodey in g:
          if(nodex != nodey):
              if(max(layout[nodex][1],layout[nodey][1])-min(layout[nodex][1],layout[nodey][1]) <0.6):
                  if((max(layout[nodex][0],layout[nodey][0])-min(layout[nodex][0],layout[nodey][0])<0.3)):
                      if(layout[nodex][1] < layout[nodey][1]):
                          layout[nodex][1] = layout[nodex][1]-0.7
                          layout[nodey][1] = layout[nodey][1]+0.7
                      else:
                          layout[nodex][1] = layout[nodex][1]+0.7
                          layout[nodey][1] = layout[nodey][1]-0.7

plt.axis('off')

plt.title("Keywords")

plt.savefig('keywords_display.png')
plt.show()
```
