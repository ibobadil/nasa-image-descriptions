```python
import csv
import pandas as pd
df = pd.read_csv('nasa_data_keywords.csv')

data = df.sort_values('date_created')
data.value_counts('date_created').plot(kind = 'bar', xlabel = 'Date', ylabel = 'Count', title = 'Image and Video Creation Dates')
```

```python
data.value_counts('keywords').nlargest(25).plot(kind = 'barh', xlabel = 'Count', ylabel = 'Keywords', title = '25 Most Frequent Keywords')
```

```python
data.value_counts('title').nlargest(15).plot(kind = 'barh', xlabel = 'Title', ylabel = 'Count', title = '15 Most Frequent Titles')
```

```python
data.value_counts('media_type').plot(kind = 'bar', xlabel = 'Media Type', ylabel = 'Count', title = 'Media Types')
```
